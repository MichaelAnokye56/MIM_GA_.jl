```julia
#| export 

# The unified, lightweight lazy proxy struct
struct SeafloorMap <: AbstractSeafloorMap
    path::String
    
    function SeafloorMap(path::AbstractString)
        if !isfile(path)
            throw(SystemError("Seafloor data file not found at: $path"))
        end
        return new(String(path))
    end
end
```
